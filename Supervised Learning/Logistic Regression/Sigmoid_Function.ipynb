```python
import numpy as np
import pandas as pd
import math 
```

```python
def Sigmoid(z):
    return 1/(1+math.exp(-z))

z = 0.95 
y_hat = Sigmoid(z)
print(y_hat)
```

```python
def y_predict(w, x, b):
    y = w*x + b 
    y_hat = Sigmoid(y)
    return y_hat 

def compute_loss(w, x, b):
    y = w*x+b 
    y_hat = y_predict(w,x,b)
    loss = -y*math.log(y_hat)-(1-y)*math.log(1-y_hat)
    return compute_loss

def dao_ham(w, x, b):
    dao_ham_w = x*(y_predict(w, x, b) - (w*x + b))
    dao_ham_b = y_predict(w, x, b) - (w*x + b)
    return dao_ham_w, dao_ham_b

def update(w, x, b, learning_rate):
    dw, db = dao_ham(w, x, b)
    new_w = w - dw 
    new_b = b - db
    return new_w, new_b

w = 0.4 
x = 3
b = -2 
learning_rate = 0.01

print(f"Loss = {compute_loss(w,x,b)}")
print(dao_ham(w, x, b))
print(f"Update training: {update(w, x, b, learning_rate)}")
```
